A comparação pelo texto exato do nome exagera as diferenças: várias colunas são o mesmo indicador escrito de outro jeito (maiúsculas, plural, espaço sobrando no fim do nome). Para separar **renomeação** de **mudança real**, normalizamos os nomes (minúsculas, sem acentos, sem espaços nas pontas).

```python
# normalizar() está em src/ips_rmbh/dados.py
conjuntos = {'2024': set(df24.columns), '2025': set(df25.columns), '2026': set(df26.columns)}
todas = sorted(set().union(*conjuntos.values()))

linhas = []
for c in todas:
    presente = [c in conjuntos[ano] for ano in conjuntos]
    linhas.append({
        'coluna': c,
        'nome_normalizado': normalizar(c),
        '2024': '✓' if presente[0] else 'X',
        '2025': '✓' if presente[1] else 'X',
        '2026': '✓' if presente[2] else 'X',
        'em_quantas': sum(presente),
    })

tabela = pd.DataFrame(linhas)
diferencas = tabela[tabela['em_quantas'] < 3].sort_values('nome_normalizado')
diferencas
```

```python
n24, n25, n26 = ({normalizar(c) for c in d.columns} for d in (df24, df25, df26))

print(f"Colunas em comum nas 3 edições -> nome exato: {len(colunas_comuns)} | nome normalizado: {len(n24 & n25 & n26)}")
print()
print("Saíram de 2024 para 2025:", sorted(n24 - n25))
print("Entraram em 2025:", sorted(n25 - n24))
print()
print("Diferenças de 2025 para 2026 (após normalizar):", sorted(n25 ^ n26))
```

```python
nulos_por_ano = pd.DataFrame({
    ano: d.rename(columns=normalizar).isna().sum()
    for ano, d in [('2024', df24), ('2025', df25), ('2026', df26)]
})
nulos_por_ano[(nulos_por_ano > 0).any(axis=1)]
```

### Conclusão da exploração

- **De 2024 para 2025, o conjunto de indicadores mudou:** saiu *Trabalho Infantil*; entraram *Famílias em Situação de Rua*, *IVCAD*, *Consumo de Ultraprocessados*, *Resposta a Processos Previdenciários* e *Resposta a Processos Familiares*; e o Enem passou de **nota média** para **nota mediana**, uma estatística diferente.
- **De 2025 para 2026, os indicadores são os mesmos**, com grafias diferentes. Mas o tratamento mudou: *Violência Contra Indígenas* passou de 745 para 4.078 valores faltantes.
- **A população de referência também muda** a cada edição (2022, 2024 e 2025).

Como o IPS e seus componentes são calculados a partir desses indicadores, as notas de edições diferentes não medem exatamente a mesma coisa. Por isso, **as perguntas de negócio usam apenas a edição 2026**, sem comparar notas entre anos.
